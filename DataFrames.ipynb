```python
import pandas as pd
import numpy as np
```

```python
df1=pd.DataFrame({'name':['harry potter','david baker','john smith','juan martinez','jane connor'],'gender':['male','male','male','male','female'],'age':[23,31,22,36,30]})
```

```python
df1
```

```python
df2=pd.DataFrame({'name':['john smith','alex du bois','joanne rowling','jane connor'],'position':['intern','team lead','manager','manager'],'wage':[2500,75000,90000,70000]})
```

```python
df2
```

```python
pd.merge(df1,df2,on="name")
```

```python
#if we give different name then use left on and right on
pd.merge(df1,df2,left_on='name',right_on='name',how='inner')
```

```python
pd.merge(df1,df2,how='left')
```

```python
pd.merge(df1,df2,how='right')
```

```python
pd.merge(df1,df2,how='outer')
```

```python
help(pd.merge)
```

```python
help(pd.concat)
```

```python
pd.concat([df1,df2],sort=True)
```

```python
pd.concat([df1,df2],axis=1,sort=True)
```

```python
df3=pd.DataFrame({'a':['a0','a1','a2','a3'],'b':['b0','b1','b2','b3'],'c':['c0','c1','c2','c3']},index=[0,1,2,3])
df3
```

```python
df4=pd.DataFrame({'a':['a2','a3','a4','a5'],'b':['b2','b3','b4','b5'],'c':['c2','c3','c4','c5'],'d':['d2','d3','d4','d5']},index=[2,3,4,5])
df4
```

```python
pd.concat([df3,df4])
```
