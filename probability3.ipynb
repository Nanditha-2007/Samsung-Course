```python
import scipy.stats as st
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

# 1.1 Binomial


```python
#sample size and p parameter
n=10
p=0.5
st.binom.pmf(5,n,p)

```

```python

#quantile at alpha 0.3
st.binom.ppf(0.3,n,p)
```

```python
#p(0<=X<=5)
st.binom.cdf(5,n,p)
```

```python
#p(3<=X<=7)
st.binom.cdf(7,n,p)-st.binom.cdf(2,n,p)
```

```python
x=np.arange(0.11)
plt.scatter(x,st.binom.pmf(x,n,p),color='red')
plt.show()
```

```python
#poissons distribution
lamb=2
```

```python
st.poisson.pmf(2,lamb)
```

```python
#quantile at alpha =0.5
#more abut quantities int ht unit 111
st.poisson.pmf(0.5,lamb)
```

```python
st.poisson.cdf(5,lamb)
```

```python
#P(3<=x<=7)
st.poisson.cdf(7,lamb)-st.poisson.cdf(2,lamb)
```

```python
#visualizing probability distrivution
x=np.arange(0,11)
plt.scatter(x,st.poisson.pmf(x,lamb),color='green')
plt.show()
```
