```python
import numpy as np
```

```python
# Training data (AND gate)
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])
# Target output
y = np.array([0, 0, 0, 1])
# Initialize weights and bias
w = np.zeros(2)
b = 0
# Learning rate
lr = 1
# Training
for epoch in range(10):
    print("Epoch:", epoch + 1)
    for i in range(len(X)):
        # Calculate weighted sum
        z = np.dot(X[i], w) + b
        # Step activation function
        if z >= 0:
            prediction = 1
        else:
            prediction = 0
        # Calculate error
        error = y[i] - prediction
        # Update weights
        w = w + lr * error * X[i]
        # Update bias
        b = b + lr * error
        print("Input:", X[i],
              "Target:", y[i],
              "Prediction:", prediction)
print("\nFinal weights:", w)
print("Final bias:", b)
```
