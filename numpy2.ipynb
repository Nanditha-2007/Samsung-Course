```python
import numpy as np
a=np.array([1,2,3])
b=np.array([[1,2],[3,4]])
print(a[0])
print(b[1,0])
print(a[[0,1,2]])          #when you want to get mul elements in 1d
print(b[[0,0],[0,1]])
print(a[a>1])
print("\n")
print(b[:,:1])
print(b[:,0])
```
