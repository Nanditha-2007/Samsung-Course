```python
z=np.cos(x)
fig0=plt.figure(figsize=(5,2),dpi=100) #,width,height and dpi setting
axes0=fig0.add_axes([0,0,1,1])
axes0.plot(x,y,color='orange',linestyle='--',label='sin')
axes0.plot(x,z,color='purple',linestyle='--',label='cos')
axes0.legend(loc=0)
axes0.set_xlabel('x')
axes0.set_ylabel('Y')
axes0.set_title('LINE PLOT')

plt.show()
```

```python
fig0=plt.figure(figsize=(8,3),dpi=80)
axes1=fig0.add_axes([0,0,0.3,1])       #left ,bottom,width,height
axes2=fig0.add_axes([0.5,0,0.3,1])
axes1.plot(x,y,color='orange',linestyle='--')
axes2.plot(x,z,color='purple',linestyle='--')
axes1.set_xlabel('x')
axes1.set_ylabel('Y')
axes1.set_title('Sine')
axes2.set_xlabel('x')
axes2.set_ylabel('Y')
axes2.set_title('Cosine')
plt.show()
```

```python
df.loc[:,'PetalLengthCm'].plot.hist(color='orange',alpha=0.5)
plt.show()
```

```python
df.loc[:,['SepalWidthCm','PetalLengthCm']].plot.hist(bins=50,color=['red','orange'],alpha=0.5)
plt.show()
```

```python
frequencies=df.Species.value_counts()
my_counts=list(frequencies.values)
my_labels=list(frequencies.index)
df2=pd.DataFrame({'counts':my_counts},index=my_labels)
df2
```

```python
df.plot.scatter(x='PetalLengthCm',y='PetalWidthCm',color='blue',alpha=0.5,marker='o')
plt.show()
```
